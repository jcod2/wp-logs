```python
import subprocess
```

```python
# test job from jupyter
result = subprocess.run(
    ['sbatch', '--wrap', 'hostname && nvidia-smi', '--job-name', 'test'],
    capture_output=True, text=True
)

print(result.stdout)
print(result.stderr)
```

```python
result = subprocess.run(
    ['sbatch', '--wrap', 'srun hostname && srun nvidia-smi', 
     '--job-name', 'test',
     '--nodes', '2',
     '--ntasks-per-node', '1'],
    capture_output=True, text=True
)
print(result.stdout)
print(result.stderr)
```

```python
result = subprocess.run(
    ['srun', '-N', '2', '--oversubscribe', 'hostname'],
    capture_output=True, text=True
)
print(result.stdout)
print(result.stderr)
```

```python
# check job status
result = subprocess.run(['squeue'], capture_output=True, text=True)
print(result.stdout)
```

```python
# submitit test
import submitit

executor = submitit.SlurmExecutor(folder='/home/admin/slurm-logs')
executor.update_parameters(
    partition='debug',
    gres='gpu:gtx1080:1',
    mem=4,
    cpus_per_task=2,
    time=2
)

def my_job():
    import torch
    return torch.cuda.get_device_name(0)

job = executor.submit(my_job)
print(job.result())
```
