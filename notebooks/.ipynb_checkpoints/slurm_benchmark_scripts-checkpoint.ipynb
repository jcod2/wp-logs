```python
import os, subprocess

SUBMIT_DIR = "/home/admin/cluster_benchmark"
SLURM_DIR  = os.path.join(SUBMIT_DIR, "slurm")
OUT_DIR    = os.path.join(SLURM_DIR, "output")
ERR_DIR    = os.path.join(SLURM_DIR, "err")

os.makedirs(SLURM_DIR, exist_ok=True)
os.makedirs(OUT_DIR, exist_ok=True)
os.makedirs(ERR_DIR, exist_ok=True)

scripts = {
    "cpu_single":   ("scripts/cpu_single.py",   "1", "512M",  36, ""),
    "cpu_parallel": ("scripts/cpu_parallel.py",  "1", "512M",  36, ""),
    "io":           ("scripts/io_test.py",        "1", "512M",  36, ""),
    "memory":       ("scripts/memory_stress.py",  "1", "2G",    36, ""),
    "stress": ("scripts/stress_test.py", "1", "1G", 36, "#SBATCH --signal=TERM@60\n#SBATCH --time=12:00:00"),
}

template = """#!/bin/bash
#SBATCH --job-name={name}
#SBATCH --output={out_dir}/{name}_%A_%a.out
#SBATCH --error={err_dir}/{name}_%A_%a.err
#SBATCH --ntasks=1
#SBATCH --cpus-per-task={cpus}
#SBATCH --mem={mem}
#SBATCH --time=01:00:00
#SBATCH --array=1-{array}
{extra}
cd {submit_dir}
python3 {script}
"""

for name, (script, cpus, mem, array, extra) in scripts.items():
    content = template.format(
        name=name, script=script, cpus=cpus,
        mem=mem, array=array, extra=extra,
        submit_dir=SUBMIT_DIR,
        out_dir=OUT_DIR,
        err_dir=ERR_DIR,
    )
    path = os.path.join(SLURM_DIR, f"{name}.sh")
    with open(path, "w") as f:
        f.write(content)
    print(f"Written: {path}")
```

```python
for name in scripts:
    path = os.path.join(SLURM_DIR, f"{name}.sh")
    r = subprocess.run(["sbatch", path], capture_output=True, text=True, cwd=SUBMIT_DIR)
    print(f"{name}: {r.stdout.strip() or r.stderr.strip()}")
```

```python
import subprocess
import numpy as np
import matplotlib.pyplot as plt
from datetime import datetime

def parse_sacct(job_id):
    r = subprocess.run([
        "sacct", "-j", str(job_id),
        "--format=JobID,JobName,NodeList,Start,End,Elapsed,MaxRSS,State",
        "--noheader", "--parsable2"  # pipe-delimited, no padding
    ], capture_output=True, text=True)

    fmt = '%Y-%m-%dT%H:%M:%S'
    jobs = []
    for line in r.stdout.strip().splitlines():
        parts = line.split('|')
        if len(parts) < 8:
            continue
        job_id_str, name, node, start, end, elapsed, mem, state = parts[:8]

        # skip .batch and .extern lines, only keep array tasks
        if '.' in job_id_str or state != 'COMPLETED':
            continue

        try:
            task_id = int(job_id_str.split('_')[1])
            start_dt = datetime.strptime(start, fmt)
            end_dt   = datetime.strptime(end, fmt)
            mem_mb   = int(mem.replace('K','')) / 1024 if mem else 0
            jobs.append({
                'id':      task_id,
                'node':    node,
                'start':   start_dt,
                'end':     end_dt,
                'runtime': (end_dt - start_dt).total_seconds(),
                'mem_mb':  mem_mb,
            })
        except (ValueError, IndexError):
            continue

    jobs.sort(key=lambda x: x['id'])

    # queue wait = start - earliest start across all tasks
    if jobs:
        submit_time = min(j['start'] for j in jobs)
        for j in jobs:
            j['wait'] = (j['start'] - submit_time).total_seconds()

    return jobs


def plot_slurm(job_id):
    jobs = parse_sacct(job_id)
    if not jobs:
        print("No completed jobs found.")
        return

    node_colors = {'slurm01': '#378ADD', 'slurm02': '#D85A30', 'slurm03': '#1D9E75'}

    ids     = [j['id'] for j in jobs]
    runtime = [j['runtime'] for j in jobs]
    wait    = [j['wait'] for j in jobs]
    mem     = [j['mem_mb'] for j in jobs]
    nodes   = [j['node'] for j in jobs]

    fig, (ax1, ax2, ax3) = plt.subplots(3, 1, figsize=(12, 9), sharex=True)
    fig.suptitle(f"SLURM cpu_single — job {job_id}  ({len(jobs)} tasks)", fontsize=13, fontweight="bold")

    for j in jobs:
        c = node_colors.get(j['node'], '#888780')
        ax1.bar(j['id'], j['runtime'], color=c, width=0.7)
    ax1.axhline(np.mean(runtime), color='black', linewidth=0.8, linestyle=':',
                label=f"mean {np.mean(runtime):.0f}s")
    ax1.set_ylabel("Runtime (s)")
    ax1.legend(fontsize=9)
    ax1.grid(axis='y', linestyle='--', alpha=0.4)

    ax2.bar(ids, wait, color='#888780', width=0.7)
    ax2.set_ylabel("Queue wait (s)")
    ax2.grid(axis='y', linestyle='--', alpha=0.4)

    ax3.bar(ids, mem, color='#7F77DD', width=0.7)
    ax3.set_ylabel("Memory used (MB)")
    ax3.set_xlabel("Task ID")
    ax3.grid(axis='y', linestyle='--', alpha=0.4)

    # legend for nodes
    for node, color in node_colors.items():
        if node in nodes:
            ax1.bar(0, 0, color=color, label=node)
    ax1.legend(fontsize=9)

    plt.tight_layout()
    plt.savefig(f"slurm_cpu_single_{job_id}.png", dpi=150, bbox_inches="tight")
    plt.show()

    print(f"\nSummary ({len(jobs)} jobs):")
    print(f"  Runtime:    mean={np.mean(runtime):.1f}s  min={np.min(runtime):.0f}s  max={np.max(runtime):.0f}s  std={np.std(runtime):.1f}s")
    print(f"  Queue wait: mean={np.mean(wait):.1f}s  max={np.max(wait):.0f}s")
    print(f"  Memory:     mean={np.mean(mem):.1f}MB  max={np.max(mem):.1f}MB")

# usage:
plot_slurm(138)
```
